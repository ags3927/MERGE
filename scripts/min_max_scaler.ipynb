```python
import os

import numpy as np

from glob import glob
from tqdm import tqdm
from pathlib import Path
from sklearn.preprocessing import MinMaxScaler
```

```python
outdir = 'PATH_TO_STORE_MIN_MAX_SCALED_DATA'  # Replace with your actual output directory
Path(outdir).mkdir(exist_ok=True, parents=True)
```

```python
eightn_counts_paths = glob('PATH_TO_8N_SMOOTHED_COUNTS/*')
spcs_counts_paths = glob('PATH_TO_SPCS_SMOOTHED_COUNTS/*')

eightn_counts_paths.sort()
spcs_counts_paths.sort()

# Make sure the file names match
for eightn_path, spcs_path in zip(eightn_counts_paths, spcs_counts_paths):
    assert Path(eightn_path).stem == Path(spcs_path).stem, "File names do not match!"

slides = [Path(p).stem for p in spcs_counts_paths]
slides.sort()
```

```python
# Iterate over the files and min-max scale the spcs counts to the 8n counts
for slide in tqdm(slides):
    eightn_counts_path = f'PATH_TO_8N_SMOOTHED_COUNTS/{slide}.npy'
    spcs_counts_path = f'PATH_TO_SPCS_SMOOTHED_COUNTS/{slide}.npy'
    
    eight_counts = np.load(eightn_counts_path)
    spcs_counts = np.load(spcs_counts_path)
    
    # Take the max and min of the 8n counts
    eightn_max = np.max(eight_counts)
    eightn_min = np.min(eight_counts)
    
    # Min-max scale the spcs counts to these values
    scaler = MinMaxScaler(feature_range=(eightn_min, eightn_max))
    rescaled_spcs_counts = scaler.fit_transform(spcs_counts)
    
    print(f"Max 8n: {eightn_max}, Min 8n: {eightn_min}")
    print(f"Max spcs: {np.max(spcs_counts)}, Min spcs: {np.min(spcs_counts)}")
    print(f"Max rescaled: {np.max(rescaled_spcs_counts)}, Min rescaled: {np.min(rescaled_spcs_counts)}")
    
    # Save the rescaled counts
    np.save(os.path.join(outdir, Path(spcs_counts_path).name), rescaled_spcs_counts)  
    
```
